```python
df_clean = df_bronze.dropDuplicates()
```

```python
from pyspark.sql.functions import col

df_clean = df_clean.filter(col("hourly").isNotNull())
```

```python
from pyspark.sql.functions import trim, lower

df_clean = df_clean.withColumn("api_status", lower(trim(col("api_status"))))
```

```python
df_clean = df_clean.fillna({
    "latitude": 0,
    "longitude": 0,
    "api_status": "unknown"
})
```

```python
from pyspark.sql.functions import explode, arrays_zip, to_timestamp

df_silver = (
    df_clean
    .withColumn(
        "zipped",
        arrays_zip(
            col("hourly.time"),
            col("hourly.temperature_2m"),
            col("hourly.relativehumidity_2m")
        )
    )
    .withColumn("exploded", explode("zipped"))
    .select(
        to_timestamp(col("exploded.time")).alias("reading_time"),
        col("exploded.temperature_2m").cast("double").alias("temperature"),
        col("exploded.relativehumidity_2m").cast("double").alias("humidity"),
        col("latitude"),
        col("longitude"),
        col("load_date"),
        col("api_status")
    )
)
```

```python
from pyspark.sql.functions import hour, when

df_silver = df_silver.withColumn(
    "peak_hour_flag",
    when(hour("reading_time").between(7, 9), 1)
    .when(hour("reading_time").between(17, 19), 1)
    .otherwise(0)
)
```

```python
df_silver = df_silver.withColumn(
    "weather_index",
    (col("temperature") * 0.7) + (col("humidity") * 0.3)
)
```

```python
df_silver.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .save("/mnt/silver/smart city/weather")

print("✔ Silver Layer Saved")
display (df_silver )
```

Silver Layer: Weather Data Pipeline

 Overview
This notebook handles the ingestion and transformation of Weather Data sourced from external APIs. The raw data arrives in complex, nested JSON format with array-based time series.

The goal is to flatten these arrays into a structured, row-based Delta Table compatible with the Energy data for joining.

🛠 Pipeline Steps (Transformation Logic)
1. Data Ingestion:

Source: Raw JSON files from /mnt/raw/smart city/weather.

Format: Multi-line JSON handling via Spark Read API.

2. Handling Nested Data (The Challenge): Raw weather data is structured as "Parallel Arrays" (e.g., one array for all timestamps, another for all temperatures).

 arrays_zip(): Aligned the corresponding elements from time, temperature, and humidity arrays into a single struct structure.

 explode(): Transformed the zipped arrays into individual rows (Row-based Time Series), creating a proper historical record.

3. Data Cleaning & Type Casting:

Converted ISO strings to TimestampType (timestamp_converted).

Cast numerical values (temperature, humidity) to DoubleType.

Renamed columns to business-friendly names:

temperature_2m → feels_like

relativehumidity_2m → humidity_index

4. Storage & Output:

Destination: /mnt/silver/smart city/weather

Format: Delta Lake.

Schema Evolution: Enabled overwriteSchema to accommodate the flattened structure.
