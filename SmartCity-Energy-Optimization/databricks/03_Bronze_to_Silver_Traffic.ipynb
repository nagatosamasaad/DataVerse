```python
from pyspark.sql.functions import current_timestamp, lit, when, col

df_clean = (
    df_clean
        .withColumnRenamed("load_date ", "load_date")
        .withColumn("ingest_time",
                    when(col("ingest_time").isNull(), current_timestamp())
                    .otherwise(col("ingest_time")))
        .withColumn("source",
                    when(col("source").isNull(), lit("unknown"))
                    .otherwise(col("source")))
        .fillna({"load_date": "none"})  
)
```

```python
from pyspark.sql.functions import lower, trim, col

df_clean = df_clean.withColumn(
    "source",
    lower(trim(col("source")))
)
```

```python
import pyspark.sql.functions as F
df_clean = df_clean.withColumn("source", F.lower(F.trim(F.col("source"))))
```

```python
df_clean = df_clean.filter(col("flowSegmentData").isNotNull())

print(" Cleaning done")
display(df_clean.limit(3))
```

```python
from pyspark.sql.functions import (
    col, when, lit, round, current_timestamp,
    to_timestamp, lower, trim
)

df_silver = (
    df_clean
    .select(
        col("load_date").alias("raw_timestamp"),
        col("flowSegmentData.currentSpeed").alias("current_speed"),
        col("flowSegmentData.freeFlowSpeed").alias("free_flow_speed"),
        col("ingest_time"),
        col("source")
    )
    .withColumn("timestamp", to_timestamp(col("raw_timestamp")))
    .withColumn("current_speed", col("current_speed").cast("double"))
    .withColumn("free_flow_speed", col("free_flow_speed").cast("double"))
    .withColumn(
        "congestion_index",
        when(
            (col("free_flow_speed").isNotNull()) & (col("free_flow_speed") != 0),
            round((col("free_flow_speed") - col("current_speed")) / col("free_flow_speed"), 2)
        ).otherwise(lit(None))
    )
    .withColumn(
        "traffic_status",
        when(col("congestion_index") >= 0.6, lit("Heavy"))
        .when(col("congestion_index") >= 0.3, lit("Moderate"))
        .when(col("congestion_index").isNotNull(), lit("Free Flow"))
        .otherwise(lit("Unknown"))
    )
    .withColumn("processing_time", current_timestamp())
)
```

```python
from pyspark.sql.functions import (hour)
df_silver = df_silver.withColumn(
    "peak_hour_flag",
    when(hour("timestamp").between(7,9), 1)
    .when(hour("timestamp").between(17,19), 1)
    .otherwise(0)
)
```

```python
from pyspark.sql.functions import (dayofweek)
df_silver = df_silver.withColumn(
    "weekend_flag",
    when(dayofweek("timestamp").isin(1,7), 1).otherwise(0) )
```

```python
from pyspark.sql.functions import col, when, lit

if "region_id" not in df_silver.columns:
    df_silver = df_silver.withColumn("region_id", lit("Cairo"))

final_cols = [
    "timestamp", "region_id" , "source", "current_speed", "free_flow_speed",
    "congestion_index", "traffic_status", "peak_hour_flag", "weekend_flag",
     "processing_time"
]

save_path = "/mnt/silver/smart city/traffic"

df_silver.select(final_cols) \
    .write.format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .save(save_path)

print(f"Silver saved to {save_path}")

display(spark.read.format("delta").load(save_path).limit(5))
```
